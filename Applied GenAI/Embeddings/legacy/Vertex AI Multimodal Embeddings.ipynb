# Vertex AI Multimodal Embeddings

Embeddings are condensed representations of data that retain the information from the data.  They are represented by a vector of numbers, usually floats, like `[0, 1, 0, 1]` or `[0.63475, 0.234, ..., 0.2646]`.  They are a trained/learned representation for data like text, images, and tables and served as preditions form these models. Based on the type of model that is used in training they can be very good at retaining latent information like semantic meaning in text, objects in images, and correlation in tables.

[Vertex AI APIs for embeddings](https://cloud.google.com/vertex-ai/generative-ai/docs/embeddings) provide simplified access to [Google's embedding models](https://cloud.google.com/vertex-ai/generative-ai/docs/learn/models#models) hosted on Vertex AI.

This workflow will interact with the API for single, multiple, and large volumes of request for **multimodal** content: text, images, and text+image combinations. The API works with text, images, and video.  

**Multimodal**

Embeddings tools that can handle both text and images are important for embedding both in the same semantic space.  Think about embedding the word "cat" and a picture of a cat.  It is helpful to match either the text or image when the desired retrieval is information about "cats".  

For a similar workflow explorting **text embeddings** specifically check out the companion workflow:
- [Vetex AI Text Embeddings API](./Vertex%20AI%20Text%20Embeddings%20API.ipynb)


**References:**
- [Get Multimodal Embeddings](https://cloud.google.com/vertex-ai/generative-ai/docs/embeddings/get-multimodal-embeddings)
- [Multimodal Embeddings API](https://cloud.google.com/vertex-ai/generative-ai/docs/model-reference/multimodal-embeddings-api)

## Embedding Dimension Size

The default dimension size for multimodal embeddings is 1408 as show in the examples above.  The dimension can be configured to these smaller dimension sizes: 128, 256, 512.

Why would a smaller embedding be helpful?  Efficiency in storage size and lower latency in usage.  A lower dimensionn can lead to a lower accuracy for the tasks you are using embeddings for - like semantic similarity matching.

Each request can set the `dimension` parameter to the desired dimension size.

```python
response = embedder.get_embeddings(contextual_text = text, image = image, dimension = 128)
```

```python
len(response.text_embedding), len(response.image_embedding)
```

## Augment Data Source With Embeddings

Add the embedding for text and image to the data source structure.

```python
for group in breed_data:
    for breed in group['breeds']:
        text = f"{group['name']} - {breed['name']}"
        with open(local_dir+'/'+breed['image'], 'rb') as f: image_bytes = f.read()
        image = vertexai.vision_models.Image(image_bytes = image_bytes)
        embed = embedder.get_embeddings(contextual_text = text, image = image)
        breed['text_embedding'] = embed.text_embedding
        breed['image_embedding'] = embed.image_embedding
        breed['image_string'] = base64.b64encode(image_bytes).decode('utf-8')
```

## Matching

Embeddings can be used with math to measure similarity.  For deeper details into this checkout the companion workflow here: [The Math of Similarity](../The%20Math%20of%20Similarity.ipynb).  Retrieval systems handle the storage and math of similarity as a service.  For an overview of Google Cloud based solutions to retrieval check out [this companion series](../../Retrieval/readme.md).

The content below motivates multimodal retrieval with the embeddings generated above using a local vector database with brute force matching using Numpy!

### Create Local Embeddings DB - With Numpy

The data is already stored in the dictionary structure of `breed_data`.  This section builds a vertical list of the location and embeddings info: [breed group index, breed index, text embedding, image embedding].  Then Numpy arrays are constructed from the text, image and combined text+image embeddings.

```python
vector_lookup = []
for g, group in enumerate(breed_data):
    for b, breed in enumerate(group['breeds']):
        vector_lookup.append([g, b, breed['text_embedding'], breed['image_embedding']])
```

```python
text_db = np.array([row[2] for row in vector_lookup])
image_db = np.array([row[3] for row in vector_lookup])
```

```python
text_db.shape
```

```python
image_db.shape
```

### Pick A Query Embedding

Select one of the breeds as the query breed and store the image and text embeddings:

```python
query_group = 0
query_breed = 0
query_text_embedding = np.array(breed_data[query_group]['breeds'][query_breed]['text_embedding'])
query_image_embedding = np.array(breed_data[query_group]['breeds'][query_breed]['image_embedding'])
breed_data[query_group]['name'], breed_data[query_group]['breeds'][query_breed]['name']
```

### Build A Function To Compare Query To Matches

```python
def compare(matches):
    def get_image(g, b):
        image_string = breed_data[g]['breeds'][b]['image_string']
        return PIL.Image.open(io.BytesIO(base64.b64decode(image_string)))
    
    fig, ax = plt.subplots(2, 5)
    for row in range(2):
        for col in range(5):
            ax[row, col].axis('off')
            ax[row, col].set_aspect('equal')
            if row==0 and col==0:
                ax[row, col].imshow(get_image(query_group, query_breed))
                ax[row, col].set_title(f"{breed_data[query_group]['name']}\n{breed_data[query_group]['breeds'][query_breed]['name']}")
            elif row==1:
                g = vector_lookup[matches[col][0]][0]
                b = vector_lookup[matches[col][0]][1]
                ax[row, col].imshow(get_image(g, b))
                ax[row, col].set_title(f"Match {col+1}\nSimilarity = {matches[col][1]:.2f}\n{breed_data[g]['name']}\n{breed_data[g]['breeds'][b]['name']}")
    fig.set_size_inches(10, 5.5)
    plt.show()
    return
```

### Get Matches With Numpy

Use dot product to calculate similarity and find matches for a query embedding.  Why dot product?  Check out the companion workflow: [The Math of Similarity](../The%20Math%20of%20Similarity.ipynb)

Display the query and top 5 matches visually for comparison of these types:
- Find the top 5 **matching text** embeddings for a **query text** embedding
- Find the top 5 **matching image** embeddings for a **query test** embedding
- Find the top 5 **matching image** embeddings for a **query image** embedding
- Find the top 5 **matching text** embeddings for a **query image** embedding

#### Query=Text To Match=Text

```python
similarity = np.dot(query_text_embedding, text_db.T)
matches = np.argsort(similarity)[::-1][:5].tolist()
matches = [(match, similarity[match]) for match in matches]
matches
```

```python
compare(matches)
```

#### Query=Text To Match=Image

```python
similarity = np.dot(query_text_embedding, image_db.T)
matches = np.argsort(similarity)[::-1][:5].tolist()
matches = [(match, similarity[match]) for match in matches]
matches
```

```python
compare(matches)
```

#### Query=Image To Match=Image

```python
similarity = np.dot(query_image_embedding, image_db.T)
matches = np.argsort(similarity)[::-1][:5].tolist()
matches = [(match, similarity[match]) for match in matches]
matches
```

```python
compare(matches)
```

#### Query=Image To Match=Text

```python
similarity = np.dot(query_image_embedding, text_db.T)
matches = np.argsort(similarity)[::-1][:5].tolist()
matches = [(match, similarity[match]) for match in matches]
matches
```

```python
compare(matches)
```
