# BigQuery AI Functions — Overview

This notebook gives you a quick tour of BigQuery's AI functions — one working example per category so you can see what's available and decide where to dive deeper.

**Categories covered:**
1. **Generation** — Generate text and structured output with `AI.GENERATE`
2. **Managed Functions** — Classify, score, and filter with `AI.CLASSIFY`, `AI.SCORE`, `AI.IF`
3. **Structured Output** — Generate tabular data with `AI.GENERATE_TABLE`
4. **Embeddings & Search** — Create embeddings with `AI.EMBED` and search with `VECTOR_SEARCH`
5. **Predictive AI** — Forecast time series with `AI.FORECAST` and predict tabular outcomes with `AI.PREDICT`
6. **Augmented Analytics** — Explain what moved a metric with `AI.KEY_DRIVERS`
7. **Utility** — Size and cost prompts before sending them with `AI.COUNT_TOKENS`

**For deeper examples** of any function, see the individual function notebooks in the [`functions/`](functions/) folder.

## 1. Generation — `AI.GENERATE`

`AI.GENERATE` is the simplest way to call a Gemini model from SQL. It's a **scalar function** — use it in SELECT, WHERE, or anywhere you'd use a regular function. No model creation required.

**Deep dive:** [`functions/ai_generate/`](functions/ai_generate/)

```python
%%bigquery --project {PROJECT_ID}

-- Simplest possible call: just a prompt string
SELECT
  city,
  (AI.GENERATE(CONCAT('What country is ', city, ' in? Answer in one word.'))).result AS country
FROM UNNEST(['Tokyo', 'Paris', 'Nairobi', 'Lima']) AS city
```

### Structured output with `output_schema`

`AI.GENERATE` can return typed columns instead of free-form text when you provide an `output_schema`.

```python
%%bigquery --project {PROJECT_ID}

-- Structured output: get typed columns back
SELECT
  animal,
  result.habitat,
  result.average_lifespan_years,
  result.is_endangered
FROM
  UNNEST(['Eagle', 'Salmon', 'Cobra']) AS animal,
  UNNEST([
    AI.GENERATE(
      CONCAT('Give me facts about: ', animal),
      output_schema => 'habitat STRING, average_lifespan_years INT64, is_endangered BOOL'
    )
  ]) AS result
```

## 2. Managed Functions — `AI.CLASSIFY`, `AI.SCORE`, `AI.IF`

These functions provide **simplified interfaces** for common tasks. BigQuery automatically optimizes the prompts sent to the model. They return simple scalar values (STRING, FLOAT64, BOOL) — not structs.

**Deep dives:** [`functions/ai_classify/`](functions/ai_classify/) | [`functions/ai_score/`](functions/ai_score/) | [`functions/ai_if/`](functions/ai_if/)

```python
%%bigquery --project {PROJECT_ID}

-- AI.CLASSIFY: assign a category from a list you provide
-- AI.SCORE: rate on a scale you describe
-- AI.IF: evaluate a natural language condition to TRUE/FALSE
SELECT
  review,
  AI.CLASSIFY(
    review,
    ['Product Quality', 'Shipping', 'Customer Service', 'Pricing']
  ) AS category,
  AI.SCORE(
    CONCAT('Rate the sentiment of this review on a scale from 1 (very negative) to 5 (very positive): ', review)
  ) AS sentiment_score,
  AI.IF(
    CONCAT(review, ' — Does this review mention a refund request?')
  ) AS mentions_refund
FROM UNNEST([
  'The product broke after two days. I want my money back!',
  'Shipping was incredibly fast, arrived next day.',
  'Great price for the quality. Will buy again.',
  'Customer support was unhelpful and rude.'
]) AS review
```

## 3. Structured Output — `AI.GENERATE_TABLE`

`AI.GENERATE_TABLE` is a **table-valued function** that returns structured columns defined by an `output_schema`. It requires a connection and remote model — both are created below.

**One output row per input prompt row.** The function calls the model once for each row of the input relation and returns one row per call, so the row count comes from the input, never from the prompt text: a single prompt asking for ten records returns one row containing one record. To get ten rows, pass ten prompts — the example below drives them from `UNNEST(GENERATE_ARRAY(1, 10))`. Input columns pass through, so keep the generator column (`n` here) to order and join on.

This is particularly useful for **generating sample data** — which is how we use it throughout this project to keep examples self-contained.

**Deep dive:** [`functions/ai_generate_table/`](functions/ai_generate_table/)

```python
import subprocess, json

# Create connection (idempotent — needed for AI.GENERATE_TABLE)
subprocess.run(
    ['bq', 'mk', '--connection', '--location', LOCATION,
     '--connection_type', 'CLOUD_RESOURCE',
     '--project_id', PROJECT_ID, CONNECTION_ID],
    capture_output=True, text=True
)

# Grant Vertex AI User role to connection's service account (idempotent)
r = subprocess.run(
    ['bq', 'show', '--connection', '--format=json',
     '--project_id', PROJECT_ID, '--location', LOCATION, CONNECTION_ID],
    capture_output=True, text=True, check=True
)
sa = json.loads(r.stdout)['cloudResource']['serviceAccountId']
subprocess.run(
    ['gcloud', 'projects', 'add-iam-policy-binding', PROJECT_ID,
     f'--member=serviceAccount:{sa}', '--role=roles/aiplatform.user', '--quiet'],
    capture_output=True, text=True
)
print(f'Connection {CONNECTION_ID} ready (service account: {sa})')

# Create remote model (idempotent)
create_model_sql = f"""
CREATE OR REPLACE MODEL `{PROJECT_ID}.{DATASET_ID}.gemini_flash`
  REMOTE WITH CONNECTION `{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}`
  OPTIONS (endpoint = 'gemini-2.5-flash')
"""
client.query(create_model_sql).result()
print('Model gemini_flash ready')
```

```python
N_TICKETS = 10  # row count comes from the input relation, not from the prompt text

query = f"""
SELECT n, ticket_id, customer_name, department, ticket_text
FROM AI.GENERATE_TABLE(
  MODEL `{PROJECT_ID}.{DATASET_ID}.gemini_flash`,
  (
    -- One prompt row per ticket: N input rows in, N output rows out
    SELECT
      n,
      CONCAT(
        'Write one fictional customer support ticket for an online retailer. ',
        'It is a ', ['complaint', 'question', 'compliment'][ORDINAL(MOD(n, 3) + 1)], '. ',
        'Use ticket_id ', CAST(1000 + n AS STRING), '. ',
        'Make the text realistic and vary the length.'
      ) AS prompt
    FROM UNNEST(GENERATE_ARRAY(1, {N_TICKETS})) AS n
  ),
  STRUCT(
    'ticket_id INT64, customer_name STRING, ticket_text STRING, department STRING' AS output_schema
  )
)
ORDER BY n
"""
df_tickets = client.query(query).to_dataframe()
print(f'Prompt rows in: {N_TICKETS}   |   Generated rows out: {len(df_tickets)}')
df_tickets
```

## 4. Embeddings & Search — `AI.EMBED` + `VECTOR_SEARCH`

`AI.EMBED` creates vector embeddings from text (scalar function, no model needed). `VECTOR_SEARCH` finds the nearest neighbors from pre-computed embeddings stored in a table.

**Deep dives:** [`functions/ai_embed/`](functions/ai_embed/) | [`functions/vector_search/`](functions/vector_search/)

```python
%%bigquery --project {PROJECT_ID}

-- AI.EMBED returns a vector embedding for each input text
SELECT
  doc,
  ARRAY_LENGTH(embedding) AS dimensions
FROM (
  SELECT
    doc,
    (AI.EMBED(
      content => doc,
      endpoint => 'text-embedding-005'
    )).result AS embedding
  FROM UNNEST([
    'BigQuery is a serverless data warehouse',
    'Cloud Storage stores unstructured data as objects',
    'Vertex AI provides ML model training and serving',
    'Cloud Run deploys containerized applications',
    'Pub/Sub is a messaging service for event-driven systems'
  ]) AS doc
)
```

### Semantic search with `VECTOR_SEARCH`

`VECTOR_SEARCH` finds the closest matches from pre-computed embeddings. It requires embeddings stored in a table — create one with `AI.EMBED`, then search it.

**Hybrid search (Preview):** there is no separate hybrid search function. Hybrid — semantic similarity combined with lexical (keyword) matching — is a capability of the search functions above: pass `lexical_search_columns` and `lexical_search_query_value` to the single-query form of `VECTOR_SEARCH`, or set `mode => 'HYBRID'` on `AI.SEARCH`. The `distance` a hybrid search returns is a fused rank score rather than a cosine distance, so it is not comparable to semantic-only distances.

**How the fusion behaves.** Hybrid ranking is reciprocal rank fusion over 1-based ranks: `distance = 1 - ( 1/(60 + rank_vector) + 1/(61 + rank_lexical) )`, lowest first. The 60 is the canonical RRF constant over 1-based ranks; the lexical leg behaves as though its ranks start at 2, which is the same arithmetic as writing its base as 61. The lexical leg does not see the whole table: BigQuery hands it a candidate pool of the top `10 * top_k` rows by semantic rank, and a row deeper than that gets no lexical rank at all, however perfectly the token matches. Within that boundary the lexical leg ranks the entire candidate pool: BM25 matches take lexical ranks 1..m and every remaining pooled row falls back to its semantic order behind them, so **no pooled row forfeits a term**. Rank 1 in both legs is the best attainable score, `1 - (1/61 + 1/62) = 0.967478`. When nothing in the pool matches the lexical value, every row falls back to its semantic order in both legs, so each row at semantic rank `r` scores `1 - ( 1/(60 + r) + 1/(61 + r) )` — that is the zero-match signature, not a general rule for unmatched rows. Once any match fires, the matches take lexical ranks 1..m and each remaining row is pushed down behind them, so an unmatched row at semantic rank `r` carries `rank_lexical > r`. A lexical match promotes a row to lexical rank 1, worth `1/62` ≈ `0.0161` — enough to pull rows in from outside the semantic top-`top_k`, but only so deep. Two gates set how deep: the pool (`rank_vector <= 10 * top_k`) and the fused score (the promoted row must still beat the row holding the last slot). Effective reach is the smaller of the two — the score gate binds below `top_k => 51`, the pool gate at and above it — so a matched row can come from semantic rank 23 at `top_k => 10`, rank 110 at `top_k => 30`, rank 510 at `top_k => 51`, rank 640 at `top_k => 64`, and rank 1,000 at `top_k => 100`. So hybrid re-ranks *and* widens, but the widening is proportional to `top_k`, never unbounded: at small `top_k` an exact token only nudges rows already near the top, so do not expect recall from it; to reach a row sitting at semantic rank `R`, size `top_k` to at least `R/10`, and to more than that for targets in the first few hundred ranks, where the score gate binds instead. When the identifier is the whole query and you want certainty, use a predicate — `WHERE sku = @sku` cannot be outranked by a fusion score, and it has no pool. See [`functions/vector_search/`](functions/vector_search/), [`functions/ai_search/`](functions/ai_search/), and the [Catalog Search](workflows/catalog_search/) workflow.

```python
# Step 1: Create a table with pre-computed embeddings
embed_sql = f"""
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.overview_docs` AS
SELECT
  doc,
  (AI.EMBED(content => doc, endpoint => 'text-embedding-005')).result AS embedding
FROM UNNEST([
  'BigQuery is a serverless data warehouse',
  'Cloud Storage stores unstructured data as objects',
  'Vertex AI provides ML model training and serving',
  'Cloud Run deploys containerized applications',
  'Pub/Sub is a messaging service for event-driven systems'
]) AS doc
"""
client.query(embed_sql).result()
print('Embeddings table ready')

# Step 2: Search for the most relevant documents
search_sql = f"""
SELECT base.doc, distance
FROM VECTOR_SEARCH(
  TABLE `{PROJECT_ID}.{DATASET_ID}.overview_docs`,
  'embedding',
  (SELECT (AI.EMBED(
    content => 'How do I analyze large datasets?',
    endpoint => 'text-embedding-005'
  )).result AS embedding),
  top_k => 3,
  distance_type => 'COSINE'
)
ORDER BY distance
"""
client.query(search_sql).to_dataframe()
```

## 5. Predictive AI — `AI.FORECAST` and `AI.PREDICT`

Two zero-training foundation models cover prediction in BigQuery: **TimesFM** for time series and **TabFM** for tabular rows. Neither needs `CREATE MODEL`, a connection, or an endpoint.

`AI.FORECAST` uses Google's TimesFM foundation model to forecast time series data. No model training or creation needed — it works zero-shot on any time series.

**Pin the model version.** The TimesFM default changed from 2.0 to 2.5 with no release note, so an unpinned forecast silently returns different numbers than it used to. The call below passes `model => 'TimesFM 2.5'` explicitly; pinning is how a forecast stays reproducible.

**Deep dive:** [`functions/ai_forecast/`](functions/ai_forecast/)

```python
%%bigquery df_forecast --project {PROJECT_ID}

-- Generate a simple time series and forecast 7 days ahead
WITH daily_sales AS (
  SELECT
    DATE_ADD('2025-01-01', INTERVAL day_offset DAY) AS sale_date,
    -- Simulate sales with trend + weekly pattern + noise.
    -- The noise is a deterministic hash, not RAND(), so the plotting cell
    -- below can rebuild the exact same series that was forecast here.
    100 + (day_offset * 0.5)  -- upward trend
    + 20 * SIN(day_offset * 2 * ACOS(-1) / 7)  -- weekly cycle
    + (MOD(ABS(FARM_FINGERPRINT(CAST(day_offset AS STRING))), 1001) / 100 - 5)  -- noise in [-5, 5]
    AS daily_amount
  FROM UNNEST(GENERATE_ARRAY(0, 89)) AS day_offset
)
SELECT *
FROM AI.FORECAST(
  (SELECT * FROM daily_sales),
  data_col => 'daily_amount',
  timestamp_col => 'sale_date',
  horizon => 7,
  confidence_level => 0.95,
  model => 'TimesFM 2.5'  -- pin the version: the default flipped 2.0 -> 2.5
)
ORDER BY forecast_timestamp
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Rebuild the exact series that was forecast above (deterministic noise, same expression)
history_sql = f"""
SELECT
  DATE_ADD('2025-01-01', INTERVAL day_offset DAY) AS sale_date,
  100 + (day_offset * 0.5) + 20 * SIN(day_offset * 2 * ACOS(-1) / 7)
    + (MOD(ABS(FARM_FINGERPRINT(CAST(day_offset AS STRING))), 1001) / 100 - 5) AS daily_amount
FROM UNNEST(GENERATE_ARRAY(0, 89)) AS day_offset
ORDER BY sale_date
"""
df_history = client.query(history_sql).to_dataframe()

fig, ax = plt.subplots(figsize=(14, 5))

# History
ax.plot(df_history['sale_date'], df_history['daily_amount'],
        color='#4285F4', linewidth=1.5, label='Historical')

# Forecast
ax.plot(df_forecast['forecast_timestamp'], df_forecast['forecast_value'],
        color='#DB4437', linewidth=2, linestyle='--', label='Forecast')

# Confidence interval
ax.fill_between(
    df_forecast['forecast_timestamp'],
    df_forecast['prediction_interval_lower_bound'],
    df_forecast['prediction_interval_upper_bound'],
    color='#DB4437', alpha=0.1, label='95% Confidence Interval'
)

ax.set_title('AI.FORECAST — Zero-Shot Time Series Forecasting with TimesFM 2.5')
ax.set_xlabel('Date')
ax.set_ylabel('Daily Sales')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Tabular prediction with `AI.PREDICT`

`AI.PREDICT` (Preview) runs regression and classification on tabular rows with the built-in TabFM foundation model. Pass it training rows and rows to predict — the **type of the label column picks the task**: a numeric label gives regression, a `BOOL` or `STRING` label gives classification. Features are limited to `STRING`, `BOOL`, `INT64`, `FLOAT64`, `NUMERIC`, and `BIGNUMERIC`, up to 20 feature columns.

**Deep dive:** [`functions/ai_predict/`](functions/ai_predict/)

```python
%%bigquery --project {PROJECT_ID}

-- Regression: predict penguin body mass from the other measurements.
-- A deterministic hash splits the 342 rows into ~302 training rows and a held-out bucket.
WITH penguins AS (
  SELECT *, MOD(ABS(FARM_FINGERPRINT(TO_JSON_STRING(t))), 10) AS bucket
  FROM `bigquery-public-data.ml_datasets.penguins` AS t
  WHERE body_mass_g IS NOT NULL
)
SELECT species, island, sex, body_mass_g AS actual_body_mass_g, predicted_body_mass_g
FROM AI.PREDICT(
  -- training rows: must contain the label column
  (SELECT * EXCEPT(bucket) FROM penguins WHERE bucket < 9),
  -- rows to predict: keeping the label here lets us compare actual vs predicted
  (SELECT * EXCEPT(bucket) FROM penguins WHERE bucket = 9 LIMIT 5),
  label_col => 'body_mass_g'
)
ORDER BY species
```

## 6. Augmented Analytics — `AI.KEY_DRIVERS`

`AI.KEY_DRIVERS` performs key driver / contribution analysis: it finds the data segments that drive a change in a summable metric between an interest set and a reference set. No model, connection, or endpoint — just a table with a metric, dimensions, and a BOOL interest/reference column.

**Deep dive:** [`functions/ai_key_drivers/`](functions/ai_key_drivers/)

```python
%%bigquery --project {PROJECT_ID}

-- Why did total trip duration change from April 2016 to April 2017?
SELECT
  ARRAY_TO_STRING(drivers, ', ') AS segment,
  difference,
  contribution
FROM AI.KEY_DRIVERS(
  (SELECT
     tripduration,
     usertype,
     gender,
     (EXTRACT(YEAR FROM starttime) = 2017) AS is_interest
   FROM `bigquery-public-data.new_york_citibike.citibike_trips`
   WHERE EXTRACT(MONTH FROM starttime) = 4
     AND EXTRACT(YEAR FROM starttime) IN (2016, 2017)
     AND tripduration IS NOT NULL),
  metric_col => 'tripduration',
  dimension_cols => ['usertype', 'gender'],
  interest_label_col => 'is_interest',
  top_k => 8
)
ORDER BY contribution DESC
```
