## Setup

inputs:

```python
project = !gcloud config get-value project
PROJECT_ID = project[0]
PROJECT_ID
```

```python
REGION = 'us-central1'
SERIES = '05'
```

packages:

```python
from google.cloud import aiplatform
```

clients:

```python
aiplatform.init(project=PROJECT_ID, location=REGION)
```

```python
clients = {}
client_options = {"api_endpoint": f"{REGION}-aiplatform.googleapis.com"}
parent = f"projects/{PROJECT_ID}/locations/{REGION}"
clients['metadata'] = aiplatform.gapic.MetadataServiceClient(client_options = client_options)
```

parameters:

```python
SERVICE_ACCOUNT = !gcloud config list --format='value(core.account)' 
SERVICE_ACCOUNT = SERVICE_ACCOUNT[0]
SERVICE_ACCOUNT
```

List the service accounts current roles:

```python
!gcloud projects get-iam-policy $PROJECT_ID --filter="bindings.members:$SERVICE_ACCOUNT" --format='table(bindings.role)' --flatten="bindings[].members"
```

>Note: If the resulting list is missing [roles/storage.objectAdmin](https://cloud.google.com/storage/docs/access-control/iam-roles) then [revisit the setup notebook](../00%20-%20Setup/00%20-%20Environment%20Setup.ipynb#permissions) and add this permission to the service account with the provided instructions.

```python
metadata_stores = []
for store in clients['metadata'].list_metadata_stores(parent = parent):
    metadata_stores.append(store)
metadata_stores
```

```python
schemas = []
for schema in clients['metadata'].list_metadata_schemas(parent = metadata_stores[0].name):
    schemas.append(schema)
schemas
```

```python
artifacts = []
for artifact in clients['metadata'].list_artifacts(parent = metadata_stores[0].name):
    artifacts.append(artifact)
artifacts
```

```python
executions = []
for execution in clients['metadata'].list_executions(parent = metadata_stores[0].name):
    executions.append(execution)
executions
```

```python
contexts = []
for context in clients['metadata'].list_contexts(parent = metadata_stores[0].name):
    contexts.append(context)
contexts
```
